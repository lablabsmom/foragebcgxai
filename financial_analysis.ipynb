# SEC 10-K financial analysis

This notebook analyzes manually extracted 2023-2025 10-K figures for Microsoft, Tesla, and Apple. Values are USD millions. It is an educational input dataset for a financial chatbot, not investment advice.

```python
from pathlib import Path
import sys
import pandas as pd

PACKAGE = Path.cwd()
if not (PACKAGE / 'data').exists():
    PACKAGE = Path.cwd() / 'financial_analysis'
sys.path.insert(0, str(PACKAGE.parent))
from financial_analysis.financial_analysis import load_data, calculate_metrics, company_summary

df = load_data(PACKAGE / 'data' / 'financial_data.csv')
df
```

## Methodology

Flow metrics use the fiscal year ended in the named year. Total assets and total liabilities are the fiscal year-end balance-sheet values. Microsoft and Apple have non-calendar fiscal years. The source filings and SEC EDGAR links are documented in `README.md`.

```python
analyzed = calculate_metrics(df)
analyzed
```

## Year-over-year trends

The derived columns show percentage changes in revenue, net income, and operating cash flow. Margins and liabilities-to-assets provide additional chatbot-ready context.

```python
summary = company_summary(analyzed)
summary
```

```python
latest = analyzed[analyzed['Year'].eq(analyzed['Year'].max())]
latest[['Company', 'Net Margin (%)', 'Operating Cash Flow Margin (%)', 'Liabilities to Assets (%)']]
```

## Findings

- Microsoft has the strongest absolute growth across the period, with revenue and net income rising substantially and operating cash flow accelerating in 2024 and 2025.
- Apple grows revenue over the period and rebounds in net income in 2025; its liabilities-to-assets ratio declines in 2025 after increasing in 2024.
- Tesla's revenue is broadly flat across the three years, while net income declines materially after 2023. Operating cash flow remains positive but does not grow with the balance sheet.
- These patterns suggest chatbot responses should distinguish scale from momentum and should cite the fiscal period and metric definition alongside any comparison.
